## Findings

1. **SMS 'backfires' in observational data** (27.6% vs 16.7%): reminders are
   sent to late bookings and high-risk patients — confounding, not causation.
   A model must learn this; a policy must not conclude 'stop sending SMS'.
2. **Lead time is the master variable** (+0.18, monotonic buckets): the longer
   the wait, the colder the intent. Day 52's headline feature.
3. **Prior history is near-zero linear signal** only because most patients have
   one visit — kept anyway for the 48k repeat rows where it matters.
